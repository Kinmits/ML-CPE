*Incorrect Data Correction*

```python
import pandas as pd

# 1. โหลดข้อมูล
df = pd.read_csv('fpl.csv')


# 2. ฟังก์ชันสำหรับตรวจสอบ Data Anomalies & Logical Errors
def validate_fpl_data(data):
    issues = []

    # ตรวจสอบค่าติดลบที่ไม่ควรเกิดขึ้น
    numeric_cols = data.select_dtypes(include=['number']).columns
    for col in numeric_cols:
        if (data[col] < 0).any() and 'cost_change' not in col:
            issues.append(f"พบค่าติดลบในคอลัมน์ {col}")

    # ตรวจสอบตรรกะ: ลงเล่น 0 นาทีแต่มีคะแนน
    logical_errors = data[(data['minutes'] == 0) & (data['total_points'] > 0)]
    if len(logical_errors) > 0:
        issues.append(
            f"พบนักเตะลงเล่น 0 นาทีแต่มีคะแนนจำนวน {len(logical_errors)} คน"
        )

    # ตรวจสอบราคาผู้เล่นที่ผิดปกติ
    cost_errors = data[(data['now_cost'] < 3.5) | (data['now_cost'] > 16.0)]
    if len(cost_errors) > 0:
        issues.append(f"พบราคาผู้เล่นหลุดช่วงมาตรฐานจำนวน {len(cost_errors)} คน")

    return issues


# 3. รันการตรวจสอบ
detected_issues = validate_fpl_data(df)

if not detected_issues:
    print(
        "✅ ตรวจสอบเรียบร้อย: ไม่พบ Incorrect Data หรือข้อมูลที่ขัดต่อตรรกะใน Dataset"
    )
else:
    print("⚠️ พบข้อมูลที่ต้องแก้ไข:", detected_issues)

# 4. Standardize Data Format (Trim whitespace เพื่อป้องกันความผิดพลาด)
string_cols = df.select_dtypes(include=['object']).columns
for col in string_cols:
    df[col] = df[col].astype(str).str.strip()

print("🧹 ทำการ Standardize ข้อมูลประเภทข้อความเรียบร้อย")
```

**Part 4: Feature Engineering**

*Label Encoding*  

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder

# 1. โหลดข้อมูล
df = pd.read_csv("fpl.csv")

# 2. จัดการ Missing Values ในคอลัมน์ข้อความก่อนทำ Label Encoding
df["news"] = df["news"].fillna("No News")

# 3. ระบุคอลัมน์ประเภท Categorical/String ที่ต้องการทำ Label Encoding
categorical_cols = ["club_name", "position_name", "news"]

# 4. ดำเนินการ Label Encoding
le = LabelEncoder()
for col in categorical_cols:
    df[f"{col}_encoded"] = le.fit_transform(df[col].astype(str))

# 5. ตรวจสอบผลลัพธ์
print(
    df[
        [
            "club_name",
            "club_name_encoded",
            "position_name",
            "position_name_encoded",
        ]
    ].head()
)
```

*One-Hot Encoding*  

```python
import pandas as pd

# 1. โหลดข้อมูล
df = pd.read_csv("fpl.csv")

# 2. จัดการ Missing Values ก่อนทำ One-Hot Encoding
df["news"] = df["news"].fillna("No News")

# 3. เลือกคอลัมน์ Categorical ที่ต้องการทำ One-Hot Encoding
categorical_cols = ["position_name", "club_name"]

# 4. ดำเนินการ One-Hot Encoding
df_encoded = pd.get_dummies(
    df, columns=categorical_cols, prefix=categorical_cols, dtype=int
)

# 5. ตรวจสอบผลลัพธ์
print(f"ขนาดข้อมูลหลังทำ One-Hot Encoding: {df_encoded.shape}")
print(df_encoded.head())
```
